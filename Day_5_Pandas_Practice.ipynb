```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
    "Marks": [85, 72, 91, 64, 78],
    "Age": [20, 22, 21, 19, 23]
}

df = pd.DataFrame(data)

print(df)
print(df.head())
print(df.info())
print(df.describe())
```

```python
high_marks = df[df["Marks"] >= 80]

print(high_marks)
```

```python
young_students = df[df["Age"] < 22]

print(young_students)
```

```python
result = df[(df["Marks"] >= 70) & (df["Age"] < 22)]

print(result)
```

```python
sorted_df = df.sort_values(by="Marks", ascending=False)

print(sorted_df)
```

```python
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Marks": [85, None, 91, None],
    "Age": [20, 22, None, 19]
}

df = pd.DataFrame(data)

print(df.isnull())
print(df.isnull().sum())
```

```python
clean_df = df.dropna()
print(clean_df)
```

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)
```
