```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)


student1 = Student("Ali", 85)
student2 = Student("Sara", 92)

student1.display()
student2.display()
```

```python
class Prediction:
    def __init__(self, object_name, confidence):
        self.object_name = object_name
        self.confidence = confidence

    def show_result(self):
        print("Detected:", self.object_name)
        print("Confidence:", self.confidence)


result1 = Prediction("Plastic", 0.95)
result2 = Prediction("Metal", 0.88)

result1.show_result()
result2.show_result()
```

```python

class Student:
    # Initialize student information
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Display student information
    def display(self):
        print("Student Name:", self.name)
        print("Student Marks:", self.marks)


try:
    # Take input from the user
    name = input("Enter student name: ").strip()

    if not name:
        raise ValueError("Student name cannot be empty.")

    marks = float(input("Enter student marks (0-100): "))

    # Validate marks
    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100.")

    # Create a Student object
    student = Student(name, marks)

    # Save student information to a file
    with open("students.txt", "a", encoding="utf-8") as file:
        file.write(f"{student.name},{student.marks}\n")

    print("\nStudent record saved successfully!")

except ValueError as error:
    print("Invalid input:", error)

except OSError as error:
    print("File error:", error)


# Read and display saved student records
try:
    print("\n--- Saved Student Records ---")

    with open("students.txt", "r", encoding="utf-8") as file:
        records = file.readlines()

        if not records:
            print("No student records found.")
        else:
            for record in records:
                name, marks = record.strip().rsplit(",", 1)
                print("Student Name:", name)
                print("Student Marks:", marks)
                print("--------------------")

except FileNotFoundError:
    print("No records file exists yet. Run the program to create one.")

except OSError as error:
    print("File reading error:", error)
```
